```python
import pandas as pd
import numpy as np
import re
import os

RAW_PATH = "/content/educate_linked_challenge_dirty_assessment.xlsx"
OUT_DIR = "/content/outputs"
os.makedirs(OUT_DIR, exist_ok=True)

CLEANING_LOG = []

def log(table, field, problem, action, justification, rows_affected):
    CLEANING_LOG.append({
        "table": table, "field": field, "problem": problem,
        "action": action, "justification": justification,
        "rows_affected": rows_affected,
    })

def parse_messy_date(val):
    """Try several known formats; return NaT if unparseable."""
    if pd.isna(val):
        return pd.NaT
    s = str(val).strip()
    if s.lower() in ("not_a_date", "nan", ""):
        return pd.NaT
    for fmt in ("%Y-%m-%d", "%d/%m/%Y", "%d/%m/%y", "%B %d %Y", "%b %d %Y"):
        try:
            return pd.to_datetime(s, format=fmt)
        except ValueError:
            continue
    try:
        return pd.to_datetime(s, dayfirst=True, errors="raise")
    except Exception:
        return pd.NaT

xls = pd.ExcelFile(RAW_PATH)
print(xls.sheet_names)
```

```python
# ---------------------------------------------------------------------------
# 1. LEARNERS
# ---------------------------------------------------------------------------
learners = pd.read_excel(xls, "learners")
n0 = len(learners)

# --- 1a. Missing learner_id: cannot be used as a key, exclude from analysis ---
missing_id_mask = learners["learner_id"].isna()
n_missing_id = missing_id_mask.sum()
learners = learners[~missing_id_mask].copy()
log(
    "learners", "learner_id",
    f"{n_missing_id} rows had a blank learner_id",
    "Excluded these rows from the analytical dataset (not deleted from the raw file)",
    "learner_id is the primary key; a row with no ID cannot be joined, deduplicated, or attributed to a learner",
    n_missing_id,
)

# --- 1b. Normalise ID formatting (whitespace/case) ---
before = learners["learner_id"].copy()
learners["learner_id"] = learners["learner_id"].astype(str).str.strip().str.upper()
n_changed = (before.astype(str).str.strip().str.upper() != before).sum()
log(
    "learners", "learner_id",
    "IDs had inconsistent whitespace/case (e.g. ' LND1088' vs 'LND1088')",
    "Trimmed whitespace and standardised to uppercase",
    "Same learner must resolve to one identical key across all tables for joins to work",
    int(n_changed),
)

# --- 1c. Gender standardisation ---
gender_map = {
    "female": "Female", "f": "Female", "female ": "Female",
    "male": "Male", "m": "Male",
    "non-binary": "Non-binary",
    "unknown": "Unknown",
    "prefer not say": "Prefer not to say",
}
learners["gender"] = learners["gender"].astype(str).str.strip().str.lower().map(gender_map)
n_missing_gender = learners["gender"].isna().sum()
learners["gender"] = learners["gender"].fillna("Unknown")
log(
    "learners", "gender",
    "12 raw spellings/cases of gender plus missing values",
    "Mapped all variants to 5 standard categories; true blanks set to 'Unknown'",
    "Gender isn't used in the resource-allocation KPIs, so 'Unknown' avoids fabricating data while keeping the row usable",
    int(n_missing_gender),
)

print("Learners after ID + gender cleaning:", learners.shape)
learners.head()
```

```python
# --- 1d. Grade standardisation ---
def clean_grade(v):
    if pd.isna(v):
        return np.nan
    s = str(v).strip().lower()
    digits = re.findall(r"\d+", s)
    words = {"nine": 9, "ten": 10, "eleven": 11, "twelve": 12}
    if digits:
        n = int(digits[0])
    else:
        n = next((val for word, val in words.items() if word in s), np.nan)
    if n in (9, 10, 11, 12):
        return int(n)
    return np.nan

grade_cleaned = learners["grade"].apply(clean_grade)
raw_grade_invalid = grade_cleaned.isna().sum()
learners["grade"] = grade_cleaned
log(
    "learners", "grade",
    "9 raw formats (12, '12', 'Grade 11', 'grade12', 'G11', 'Gr 10', 'Grade Ten') for grades 9-12",
    "Extracted the numeric grade from text/number, mapped word-forms (e.g. 'Ten'->10); every value was resolvable to a valid 9-12 grade",
    "Grade must be a clean integer 9-12 to support grade-level comparisons",
    int(raw_grade_invalid),
)

# --- 1e. registration_date parsing ---
raw_dates = learners["registration_date"].copy()
learners["registration_date"] = learners["registration_date"].apply(parse_messy_date)
n_bad_dates = learners["registration_date"].isna().sum()
```

```python
# --- 1f. Duplicate learner_id resolution ---
dup_mask = learners.duplicated("learner_id", keep=False)
dup_ids = learners.loc[dup_mask, "learner_id"].unique()
n_exact_dupes = 0
n_conflict_dupes = 0
rows_to_drop = []

for lid in dup_ids:
    sub = learners[learners["learner_id"] == lid]
    if sub.nunique().eq(1).all():
        # truly identical rows -> keep first, drop the rest
        rows_to_drop.extend(sub.index[1:].tolist())
        n_exact_dupes += 1
    else:
        # conflicting: keep the row with fewer missing/placeholder values
        def completeness_score(row):
            score = 0
            if row["gender"] not in ("Unknown", None) and pd.notna(row["gender"]):
                score += 1
            if pd.notna(row["grade"]):
                score += 1
            if pd.notna(row["registration_date"]):
                score += 1
            return score

        scores = sub.apply(completeness_score, axis=1)
        keep_idx = scores.idxmax()
        drop_idx = [i for i in sub.index if i != keep_idx]
        rows_to_drop.extend(drop_idx)
        n_conflict_dupes += 1

learners = learners.drop(index=rows_to_drop)
log(
    "learners", "learner_id",
    f"{len(dup_ids)} learner_ids appeared more than once ({n_exact_dupes} exact duplicates, {n_conflict_dupes} conflicting records)",
    "Exact duplicates: kept first occurrence. Conflicting records: kept the more complete row (e.g. a real gender over 'Unknown')",
    "learner_id must be unique to serve as a join key; a documented tie-break rule is more defensible than an arbitrary keep-last",
    len(rows_to_drop),
)

learners = learners.reset_index(drop=True)
print("Final learners shape:", learners.shape)
learners.head()
```

```python
# ---------------------------------------------------------------------------
# 2. SUBJECT NAME STANDARDISATION (shared across engagement, assessments,
#    support_capacity, enquiries)
# ---------------------------------------------------------------------------
SUBJECT_MAP = {
    "maths": "Mathematics", "math": "Mathematics", "mathematics": "Mathematics",
    "sci": "Physical Sciences", "phys sci": "Physical Sciences",
    "physical science": "Physical Sciences", "physical sciences": "Physical Sciences",
    "ls": "Life Sciences", "life science": "Life Sciences", "life sciences": "Life Sciences",
    "acc": "Accounting", "accounting": "Accounting",
}

def clean_subject(v):
    if pd.isna(v):
        return np.nan
    s = str(v).strip().lower()
    return SUBJECT_MAP.get(s, s.title())
```

```python
# ---------------------------------------------------------------------------
# 3. ENGAGEMENT
# ---------------------------------------------------------------------------
engagement = pd.read_excel(xls, "engagement")
n0 = len(engagement)

engagement["learner_id"] = engagement["learner_id"].astype(str).str.strip().str.upper()

n_bad_subject = engagement["subject"].isna().sum()
engagement["subject"] = engagement["subject"].apply(clean_subject)
log(
    "engagement", "subject",
    "12 raw spellings of 4 subjects (Sci/LS/Math/'  Mathematics  ', mixed case) plus missing subject",
    "Standardised to 4 canonical names: Mathematics, Physical Sciences, Life Sciences, Accounting",
    "Subject is the key management is deciding on; inconsistent labels would silently split one subject's data into several",
    int(n_bad_subject),
)

# --- attendance %: strip '%' strings, then clear impossible values ---
def clean_pct(v):
    if pd.isna(v):
        return np.nan
    if isinstance(v, str):
        v = v.replace("%", "").strip()
    try:
        return float(v)
    except ValueError:
        return np.nan

engagement["tutoring_attendance_pct"] = engagement["tutoring_attendance_pct"].apply(clean_pct)
impossible_pct = (~engagement["tutoring_attendance_pct"].between(0, 100)) & engagement["tutoring_attendance_pct"].notna()
n_impossible_pct = impossible_pct.sum()
engagement.loc[impossible_pct, "tutoring_attendance_pct"] = np.nan
log(
    "engagement", "tutoring_attendance_pct",
    "Values stored as text with '%' sign, and impossible values above 100% (e.g. 999, 134.5)",
    "Stripped '%' and converted to numeric; values outside 0-100 set to missing (not clipped/guessed)",
    "Attendance cannot exceed 100%; clipping to 100 would misrepresent extreme data-entry errors as real perfect attendance",
    int(n_impossible_pct),
)

print("Engagement after subject + attendance cleaning:", engagement.shape)
engagement.head(10)
```

```python
# --- watch_time_minutes: European decimal comma ---
def clean_watch_time(v):
    if pd.isna(v):
        return np.nan
    if isinstance(v, str):
        v = v.replace(",", ".").strip()
    try:
        val = float(v)
        return val if val >= 0 else np.nan
    except ValueError:
        return np.nan

raw_watch = engagement["watch_time_minutes"].copy()
engagement["watch_time_minutes"] = engagement["watch_time_minutes"].apply(clean_watch_time)
n_comma = raw_watch.apply(lambda v: isinstance(v, str) and "," in v).sum()
log(
    "engagement", "watch_time_minutes",
    "Some values used a comma as decimal separator (e.g. '262,7' instead of 262.7)",
    "Replaced comma with a decimal point and converted to numeric",
    "This is a locale formatting artefact, not a real different value; treating '262,7' as the integer 2627 would badly distort averages",
    int(n_comma),
)

# --- duplicate (learner_id, subject) rows ---
dup_mask = engagement.duplicated(["learner_id", "subject"], keep="first")
n_dupe_eng = dup_mask.sum()
engagement = engagement[~dup_mask].copy()
log(
    "engagement", "(learner_id, subject)",
    f"{n_dupe_eng} rows duplicated the same learner+subject engagement record",
    "Kept the first occurrence, dropped the repeat",
    "One learner should have exactly one engagement record per subject; duplicates would double-count videos/watch time in aggregates",
    int(n_dupe_eng),
)

engagement = engagement.reset_index(drop=True)
print("Engagement final shape:", engagement.shape)
engagement["subject"].value_counts()
```

```python
# ---------------------------------------------------------------------------
# 4. ASSESSMENTS
# ---------------------------------------------------------------------------
assessments = pd.read_excel(xls, "assessments")
n0 = len(assessments)

# --- missing learner_id / subject rows: cannot attribute, exclude ---
missing_key = assessments["learner_id"].isna() | assessments["subject"].isna()
n_missing_key = missing_key.sum()
assessments = assessments[~missing_key].copy()
log(
    "assessments", "learner_id/subject",
    f"{n_missing_key} row(s) had a blank learner_id or subject",
    "Excluded from the analytical dataset (cannot attribute a mark to a learner/subject)",
    "Both fields are needed to form the join key; a blank key cannot be matched to engagement or learners",
    int(n_missing_key),
)

assessments["learner_id"] = assessments["learner_id"].astype(str).str.strip().str.upper()
assessments["subject"] = assessments["subject"].apply(clean_subject)

print("Assessments after key cleaning:", assessments.shape)
assessments.head(10)
```

```python
# --- baseline_mark: 'MISSING' string, '%' strings, and >100 outlier ---
def clean_mark(v):
    if pd.isna(v):
        return np.nan
    if isinstance(v, str):
        s = v.strip()
        if s.upper() == "MISSING":
            return np.nan
        s = s.replace("%", "")
        try:
            v = float(s)
        except ValueError:
            return np.nan
    v = float(v)
    return v if 0 <= v <= 100 else np.nan

n_baseline_text = assessments["baseline_mark"].apply(lambda v: isinstance(v, str)).sum()
assessments["baseline_mark"] = assessments["baseline_mark"].apply(clean_mark)
log(
    "assessments", "baseline_mark",
    "Values stored as literal text 'MISSING', as percentage strings ('42%'), or impossible (>100, e.g. 115)",
    "'MISSING' and out-of-range values set to NaN; '%' stripped and converted to numeric",
    "A mark cannot exceed 100; 'MISSING' is an explicit non-value, not a number to guess at",
    int(n_baseline_text),
)

# --- latest_mark: sentinel outliers (1050, 180 repeated exactly) ---
sentinel_mask = assessments["latest_mark"] > 100
n_sentinel = sentinel_mask.sum()
assessments.loc[sentinel_mask, "latest_mark"] = np.nan
log(
    "assessments", "latest_mark",
    "Rows had latest_mark > 100 with exact repeated values (1050 x9, 180 x2) - the exact repetition indicates a sentinel/placeholder code, not real measurement error",
    "Set to missing rather than guessing a 'true' value (e.g. dividing by 10)",
    "Marks are bounded 0-100; a repeated impossible constant is evidence of a system placeholder, not a transcription slip that can be safely reversed",
    int(n_sentinel),
)

# --- duplicate (learner_id, subject) rows ---
dup_mask = assessments.duplicated(["learner_id", "subject"], keep="first")
n_dupe_assess = dup_mask.sum()
assessments = assessments[~dup_mask].copy()
log(
    "assessments", "(learner_id, subject)",
    f"{n_dupe_assess} rows duplicated the same learner+subject assessment record",
    "Kept the first occurrence, dropped the repeat",
    "One learner should have one assessment record per subject; duplicates would double-count in mark-change aggregates",
    int(n_dupe_assess),
)

assessments = assessments.reset_index(drop=True)
print("Assessments final shape:", assessments.shape)
assessments["latest_mark"].describe()
```

```python
# ---------------------------------------------------------------------------
# 5. SUPPORT_CAPACITY
# ---------------------------------------------------------------------------
support = pd.read_excel(xls, "support_capacity")
n0 = len(support)
print(support)  # look at the raw mess before we touch it
```

```python
support["subject"] = support["subject"].apply(clean_subject)
log(
    "support_capacity", "subject",
    "8 rows represent only 4 real subjects, duplicated under different spellings/casing",
    "Standardised subject names to the same 4 canonical labels used elsewhere",
    "Capacity figures must be attributable to the same subject key used in engagement/assessments/enquiries, or per-subject totals silently split in half",
    n0,
)

import re

def clean_hours(v):
    if pd.isna(v):
        return np.nan
    if isinstance(v, str):
        v = re.sub(r"[^0-9.\-]", "", v)
        if v == "":
            return np.nan
    v = float(v)
    return v if v >= 0 else np.nan

def clean_cost(v):
    if pd.isna(v):
        return np.nan
    if isinstance(v, str):
        v = re.sub(r"[^0-9.]", "", v)
        if v == "":
            return np.nan
    return float(v)

n_hours_text = support["weekly_tutor_hours"].apply(lambda v: isinstance(v, str)).sum()
support["weekly_tutor_hours"] = support["weekly_tutor_hours"].apply(clean_hours)
log(
    "support_capacity", "weekly_tutor_hours",
    "Values mixed plain numbers with a unit-suffixed string ('26 hrs') and one negative value (-2, impossible)",
    "Stripped non-numeric characters, converted to numeric; negative values set to missing",
    "Hours cannot be negative; '-2' is almost certainly a sign error, and the duplicate-row merge below recovers the true value (22)",
    int(n_hours_text) + 1,
)

n_cost_text = support["estimated_cost_per_extra_hour"].apply(lambda v: isinstance(v, str)).sum()
support["estimated_cost_per_extra_hour"] = support["estimated_cost_per_extra_hour"].apply(clean_cost)
log(
    "support_capacity", "estimated_cost_per_extra_hour",
    "Values mixed a plain number, a currency-prefixed string ('R180'), and a string with trailing zeros ('175.00')",
    "Stripped non-numeric characters (currency symbols) and converted to a plain numeric Rand value",
    "All values are the same currency (ZAR); the 'R' prefix is a formatting artefact, not a different unit",
    int(n_cost_text),
)

# --- merge duplicate-subject rows: take first valid value per column ---
support = support.groupby("subject", as_index=False).agg(
    active_tutors=("active_tutors", "first"),
    weekly_tutor_hours=("weekly_tutor_hours", lambda s: s.dropna().iloc[0] if s.notna().any() else np.nan),
    waiting_list=("waiting_list", lambda s: s.dropna().iloc[0] if s.notna().any() else np.nan),
    estimated_cost_per_extra_hour=("estimated_cost_per_extra_hour", lambda s: s.dropna().iloc[0] if s.notna().any() else np.nan),
)
log(
    "support_capacity", "subject (row-level)",
    "After standardising names, each of the 4 subjects had 2 rows",
    "Merged same-subject rows by taking the first valid (non-missing, non-corrupted) value per column",
    "Support capacity is defined at subject grain; two rows per subject would double-count tutors/hours if summed, or hide the true value if the wrong row were picked",
    n0 - 4,
)

print("Support capacity final:")
support
```

```python
# ---------------------------------------------------------------------------
# 6. ENQUIRIES
# ---------------------------------------------------------------------------
enquiries = pd.read_excel(xls, "enquiries")
n0 = len(enquiries)

enquiries["subject"] = enquiries["subject"].apply(clean_subject)
log(
    "enquiries", "subject",
    "Same subject-naming inconsistency (Maths/maths, phys sci) seen in other tables",
    "Standardised to the same 4 canonical subject names",
    "Enquiry demand must roll up to the same subject key as capacity/engagement for a fair comparison",
    n0,
)

# --- grade ---
n_bad_enq_grade = 0
def clean_enq_grade(v):
    global n_bad_enq_grade
    if pd.isna(v):
        n_bad_enq_grade += 1
        return np.nan
    s = str(v).strip().lower()
    digits = re.findall(r"\d+", s)
    n = int(digits[0]) if digits else np.nan
    if n in (10, 11, 12):
        return int(n)
    n_bad_enq_grade += 1
    return np.nan

enquiries["grade"] = enquiries["grade"].apply(clean_enq_grade)
log(
    "enquiries", "grade",
    "One text variant ('Grade 12') and one impossible value (13 - Edu-ca-te only serves Grade 10-12)",
    "Extracted numeric grade from text; values outside 10-12 set to missing",
    "Grade 13 doesn't exist in the schooling system this platform serves; treating it as real would misstate the grade-demand breakdown",
    int(n_bad_enq_grade),
)

# --- enquiry_date ---
raw_enq_dates = enquiries["enquiry_date"].copy()
enquiries["enquiry_date"] = enquiries["enquiry_date"].apply(parse_messy_date)
n_bad_enq_dates = enquiries["enquiry_date"].isna().sum() - raw_enq_dates.isna().sum()
log(
    "enquiries", "enquiry_date",
    "Mixed date formats (ISO vs DD/MM/YYYY)",
    "Parsed to true datetime using the same logic as learners.registration_date",
    "Consistent dates are required to analyse whether enquiry demand is rising over the term",
    int(n_bad_enq_dates),
)

# --- converted_to_registration ---
convert_map = {"y": 1, "yes": 1, "1": 1, 1: 1, "n": 0, "no": 0, "0": 0, 0: 0}

def clean_converted(v):
    if pd.isna(v):
        return np.nan
    key = str(v).strip().lower()
    return convert_map.get(key, np.nan)  # 'maybe' and blanks stay missing

n_ambiguous_convert = enquiries["converted_to_registration"].apply(
    lambda v: pd.notna(v) and str(v).strip().lower() not in convert_map
).sum()
enquiries["converted_to_registration"] = enquiries["converted_to_registration"].apply(clean_converted)
log(
    "enquiries", "converted_to_registration",
    "8 encodings of yes/no (Y/N/Yes/No/1/0) plus a genuinely ambiguous 'Maybe' value and blanks",
    "Mapped all yes/no variants to 1/0; 'Maybe' and blanks left as missing rather than forced into a binary outcome",
    "Forcing 'Maybe' to Yes or No would fabricate a conversion outcome the source data doesn't actually confirm",
    int(n_ambiguous_convert),
)

# --- duplicate enquiry_id ---
dup_mask = enquiries.duplicated("enquiry_id", keep="first")
n_dupe_enq = dup_mask.sum()
enquiries = enquiries[~dup_mask].copy()
log(
    "enquiries", "enquiry_id",
    f"{n_dupe_enq} row(s) repeated the same enquiry_id",
    "Kept the first occurrence, dropped the repeat",
    "enquiry_id is the primary key for this table; duplicates would double-count enquiry volume",
    int(n_dupe_enq),
)

enquiries = enquiries.reset_index(drop=True)
print("Enquiries final shape:", enquiries.shape)
enquiries.head(10)
```

```python
# ---------------------------------------------------------------------------
# 7. EXPORT CLEANING LOG
# ---------------------------------------------------------------------------
log_df = pd.DataFrame(CLEANING_LOG)
log_df.to_csv(f"{OUT_DIR}/cleaning_log.csv", index=False)
print(f"Cleaning log: {len(log_df)} entries")
log_df
```

```python
# ---------------------------------------------------------------------------
# 8. JOIN INTEGRITY CHECKS
# ---------------------------------------------------------------------------
print("=== JOIN INTEGRITY CHECKS ===")

# 8a. engagement/assessments learner_ids that don't exist in cleaned learners (orphans)
learner_ids = set(learners["learner_id"])
eng_orphans = set(engagement["learner_id"]) - learner_ids
assess_orphans = set(assessments["learner_id"]) - learner_ids
print(f"engagement rows referencing unknown learner_id: {engagement['learner_id'].isin(eng_orphans).sum()} -> {sorted(eng_orphans)}")
print(f"assessments rows referencing unknown learner_id: {assessments['learner_id'].isin(assess_orphans).sum()} -> {sorted(assess_orphans)}")

log(
    "engagement/assessments", "learner_id",
    f"engagement has {len(eng_orphans)} and assessments has {len(assess_orphans)} learner_id values (e.g. LND9999, TEMP-004) with no matching row in cleaned learners",
    "Excluded these from any learner-joined analysis; kept in the standalone cleaned CSVs for transparency",
    "A learner_id with no matching learner record is a data-entry error or placeholder ID; joining it would silently invent a learner or be dropped anyway, so it's made explicit here rather than silent",
    len(eng_orphans) + len(assess_orphans),
)

# 8b. row-count check for the engagement <-> assessments join
before_left = len(engagement)
merged_check = engagement.merge(assessments, on=["learner_id", "subject"], how="inner")
print(f"engagement rows: {before_left}, assessments rows: {len(assessments)}, inner-joined rows: {len(merged_check)}")
assert len(merged_check) <= max(len(engagement), len(assessments)), "JOIN FAN-OUT DETECTED"
print("No fan-out: each (learner_id, subject) pair is unique in both tables.")
```

```python
# ---------------------------------------------------------------------------
# 9. BUILD SUBJECT-LEVEL KPI TABLE
# ---------------------------------------------------------------------------
print("=== SUBJECT-LEVEL AGGREGATION ===")

# only use engagement/assessment rows with a real learner match (excludes orphans)
engagement_valid = engagement[engagement["learner_id"].isin(learner_ids)].copy()
assessments_valid = assessments[assessments["learner_id"].isin(learner_ids)].copy()

eng_by_subject = engagement_valid.groupby("subject").agg(
    n_learners_engaged=("learner_id", "nunique"),
    avg_attendance_pct=("tutoring_attendance_pct", "mean"),
    avg_videos_watched=("videos_watched", "mean"),
    avg_watch_time_min=("watch_time_minutes", "mean"),
).reset_index()

assessments_valid["mark_change"] = assessments_valid["latest_mark"] - assessments_valid["baseline_mark"]
assess_agg = assessments_valid.groupby("subject").agg(
    n_assessed=("learner_id", "nunique"),
    avg_baseline_mark=("baseline_mark", "mean"),
    avg_latest_mark=("latest_mark", "mean"),
    avg_mark_change=("mark_change", "mean"),
).reset_index()

enquiries_by_subject = enquiries.groupby("subject").agg(
    n_enquiries=("enquiry_id", "count"),
    n_converted=("converted_to_registration", "sum"),
).reset_index()
enquiries_by_subject["conversion_rate_pct"] = (
    100 * enquiries_by_subject["n_converted"] / enquiries_by_subject["n_enquiries"]
).round(1)

kpi = (
    support.merge(eng_by_subject, on="subject", how="left")
    .merge(assess_agg, on="subject", how="left")
    .merge(enquiries_by_subject, on="subject", how="left")
)

kpi["waitinglist_per_active_tutor"] = (kpi["waiting_list"] / kpi["active_tutors"]).round(2)
kpi["demand_pressure_index"] = (
    kpi["waiting_list"].fillna(0) + kpi["n_enquiries"].fillna(0) * (kpi["conversion_rate_pct"].fillna(0) / 100)
).round(1)

kpi.to_csv(f"{OUT_DIR}/subject_kpis.csv", index=False)
kpi
```
